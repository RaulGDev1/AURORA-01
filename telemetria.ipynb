# AURORA-01 — Sistema de Telemetria

## Simulação de verificação para decolagem

Este projeto apresenta uma simulação de um sistema de telemetria
responsável por verificar as condições de segurança da nave AURORA-01
antes de sua decolagem.

Os dados utilizados são simulados para fins acadêmicos.

## 1. Dados da telemetria

O sistema monitora os seguintes parâmetros:

- Temperatura interna;
- Temperatura externa;
- Integridade estrutural;
- Nível de energia;
- Pressão dos tanques;
- Status dos módulos críticos.

Para que a decolagem seja autorizada, todos os parâmetros devem estar
dentro das condições de segurança estabelecidas.

```python
# ==========================================
# DADOS SIMULADOS DA TELEMETRIA
# ==========================================

temperatura_interna = 22
temperatura_externa = -65
integridade_estrutural = 1
energia = 87
pressao_tanques = 92
modulo_critico = "OK"

print("==========================================")
print("       TELEMETRIA DA NAVE AURORA-01")
print("==========================================")

print(f"Temperatura interna: {temperatura_interna} °C")
print(f"Temperatura externa: {temperatura_externa} °C")
print(f"Integridade estrutural: {integridade_estrutural}")
print(f"Nível de energia: {energia}%")
print(f"Pressão dos tanques: {pressao_tanques}%")
print(f"Módulo crítico: {modulo_critico}")
```

## 2. Condições de segurança

Foram definidas as seguintes condições para autorização da decolagem:

| Parâmetro | Condição segura |
|---|---|
| Temperatura interna | Entre 18 °C e 30 °C |
| Temperatura externa | Entre -120 °C e 0 °C |
| Integridade estrutural | Igual a 1 |
| Nível de energia | Maior ou igual a 70% |
| Pressão dos tanques | Maior ou igual a 80% |
| Módulo crítico | Status "OK" |

Os valores das faixas de segurança são parâmetros simulados,
definidos exclusivamente para esta atividade acadêmica.

```python
# ==========================================
# VERIFICAÇÕES DE SEGURANÇA
# ==========================================

temp_interna_ok = 18 <= temperatura_interna <= 30

temp_externa_ok = -120 <= temperatura_externa <= 0

estrutura_ok = integridade_estrutural == 1

energia_ok = energia >= 70

pressao_ok = pressao_tanques >= 80

modulo_ok = modulo_critico == "OK"

print("==========================================")
print("       VERIFICAÇÕES DE SEGURANÇA")
print("==========================================")

print(f"Temperatura interna: {'OK' if temp_interna_ok else 'FALHA'}")
print(f"Temperatura externa: {'OK' if temp_externa_ok else 'FALHA'}")
print(f"Integridade estrutural: {'OK' if estrutura_ok else 'FALHA'}")
print(f"Nível de energia: {'OK' if energia_ok else 'FALHA'}")
print(f"Pressão dos tanques: {'OK' if pressao_ok else 'FALHA'}")
print(f"Módulo crítico: {'OK' if modulo_ok else 'FALHA'}")
```

```python
# ==========================================
# DECISÃO FINAL
# ==========================================

todos_os_sistemas_ok = (
    temp_interna_ok
    and temp_externa_ok
    and estrutura_ok
    and energia_ok
    and pressao_ok
    and modulo_ok
)

print("\n==========================================")

if todos_os_sistemas_ok:
    print("PRONTO PARA DECOLAR")
else:
    print("DECOLAGEM ABORTADA")

print("==========================================")
```

simulação contendo parâmetros fora das condições de segurança.

dos tanques estarão fora dos limites estabelecidos.
